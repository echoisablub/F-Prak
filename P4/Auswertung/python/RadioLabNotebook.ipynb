There are a infinite number of options when comes to data plotting with python. A very used and simple one is `errorbar` function of matplotli.pyplot module. In the following lines you will find a simple example, if you want to modify the colours, the marker size or type, the lines of the plot etc have a look __[here](https://matplotlib.org/3.5.1/api/_as_gen/matplotlib.pyplot.errorbar.html)__. 



```python
# plot: example using plt.errorbar

fig = plt.figure(figsize=(10,7))

xval = np.arange(0.1, 4, 0.5)
yval = np.exp(-xval)
  
plt.errorbar(xval, yval, xerr = 0.4, yerr = 0.5)
  
plt.title('matplotlib.pyplot.errorbar() function Example')
plt.xlabel('')
plt.ylabel('')
# plt.savefig()
plt.show()


print('!! Remember to put the right label on both axis (with units) and errors on you data points')
```

## 2. Directly save measured data in a table using python dataframe

Now that you have you measurements, save them in different lists, as shown below. In the end they will be part of a table. Run the next lines to see an example.

```python
# intialise data of lists
mydata = {'data1': [1, 2], 
          'data2': [3, 4],
          'data3': [5,6]}

# Create DataFrame
df = pd.DataFrame(data=mydata)

# Print the output
df
```

You can access to the columns of your table use `df["column_name"]`. You will have now a list of values that you can use to calculate whatever you like (you can plot data, do operations etc).
